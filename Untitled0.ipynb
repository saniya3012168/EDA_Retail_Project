```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from IPython.display import display
plt.rcParams['figure.figsize'] = (10,5)
sns.set_style("whitegrid")

# Load dataset
DATA_PATH = "/mnt/data/product_hierarchy.csv"
if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH, low_memory=False)
    print(f"Loaded dataset from {DATA_PATH}")
else:
    # Fallback: let user upload file manually in Colab
    from google.colab import files
    uploaded = files.upload()
    # take the first uploaded file
    fname = next(iter(uploaded))
    df = pd.read_csv(fname, low_memory=False)
    print(f"Loaded uploaded file: {fname}")

print("Shape:", df.shape)
display(df.head())
```

```python
df.info()
print("\nMissing values (count per column):")
print(df.isna().sum())
```

```python
df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]
for c in df.select_dtypes(include=["object"]).columns:
    df[c] = df[c].str.strip()
print("Cleaned column names:")
print(df.columns.tolist())
```

```python
num_cols = df.select_dtypes(include=["int64","float64"]).columns.tolist()
cat_cols = df.select_dtypes(include=["object","category"]).columns.tolist()
print("Numeric columns:", num_cols)
print("Categorical columns (sample):", cat_cols[:10])
```

```python
if len(num_cols)==0:
    print("No numeric columns detected.")
else:
    mean = df[num_cols].mean()
    median = df[num_cols].median()
    mode = df[num_cols].mode().iloc[0]     # first mode if multiple
    std = df[num_cols].std()
    summary = pd.DataFrame({"mean": mean, "median": median, "mode": mode, "std": std})
    display(summary.round(4))
```

```python
display(df[num_cols].describe().transpose().round(4))
```

```python
plt.figure(figsize=(12,4))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing values (rows x cols)")
plt.show()
```

```python
if num_cols:
    df[num_cols].hist(bins=20, layout=(max(1, (len(num_cols)+2)//3),3), figsize=(12, 3*((len(num_cols)+2)//3)))
    plt.tight_layout()
    plt.show()
```

```python
if len(num_cols) >= 2:
    corr = df[num_cols].corr()
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
    plt.title("Correlation matrix (numeric)")
    plt.show()
```

```python
for c in cat_cols:
    print(f"\nTop values for `{c}`:")
    print(df[c].value_counts(dropna=False).head(5).to_string())
```

```python
if "cluster_id" in df.columns and "product_width" in df.columns:
    plt.figure(figsize=(12,5))
    sns.boxplot(x="cluster_id", y="product_width", data=df)
    plt.xticks(rotation=45)
    plt.title("product_width by cluster_id")
    plt.show()
else:
    print("Skipping example product/cluster plot (missing columns: cluster_id or product_width).")
```

```python
# Bar Charts for categorical columns
import matplotlib.pyplot as plt

cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().head(10).plot(kind='bar', color='skyblue')
    plt.title(f"Top 10 Values in {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

```python
#Line Chart for a numeric trend (choose any numeric column)
if num_cols:
    col = num_cols[0]   # first numeric column
    plt.plot(df[col].head(100), marker='o', linestyle='-', linewidth=1)
    plt.title(f"Line Trend of {col} (first 100 rows)")
    plt.xlabel("Index")
    plt.ylabel(col)
    plt.show()
else:
    print("No numeric columns found for line plot.")
```
